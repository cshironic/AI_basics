and_gate.py

```python
import numpy as np


def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = AND(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```

or_gate.py

```python
import numpy as np


def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = OR(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```

nand_gate.py

```python
import numpy as np


def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = NAND(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```

perceptron model

```python
import numpy as np

epsilon = 0.0000001  # Prevent floating-point errors


def step_func(t):  # Activation function of the perceptron
  if t > epsilon:
    return 1
  else:
    return 0


X = np.array([  # Training dataset
    [0, 0, 1],  # The last 1 is the input signal for the bias
    [0, 1, 1],  # The last 1 is the input signal for the bias
    [1, 0, 1],  # The last 1 is the input signal for the bias
    [1, 1, 1],  # The last 1 is the input signal for the bias
])
y = np.array([0, 0, 0, 1])  # NumPy matrix storing the correct answers
#y = np.array([0, 1, 1, 0])  # NumPy matrix storing the correct answers
W = np.zeros(len(X[0]))

def perceptron_fit(X, Y, epochs=10):  # Implement perceptron learning algorithm
  global W
  eta = 0.2  # Learning rate
  for t in range(epochs):
    print("epoch=", t, "======================")
    for i in range(len(X)):
      predict = step_func(np.dot(X[i], W))
      error = Y[i] - predict  # Calculate error
      W += eta * error * X[i]  # Update weights
      print("Current input=", X[i], "Answer=", Y[i], "Output=", predict,"Updated weights=",W,)
    print("================================")


def perceptron_predict(X, Y):  # Prediction
  global W
  for x in X:
    print(x[0], x[1], "->", step_func(np.dot(x, W)))


perceptron_fit(X, y, 6)
perceptron_predict(X, y)
```
